```python
Meta data
#Metadata is data about data — meaning it provides information that describes, explains, or gives context to other data.
#In simple terms, metadata helps you understand what the data is, where it comes from, and how it should be used.
Examples:
In a photo, metadata might include:
#Date and time taken
#Camera model
#Location (GPS coordinates)
In a document, metadata can include:
 #Author’s name
 #Date created or modified
 #File size and format
In a database, metadata describes:
 #Table names
 #Column types (text, number, date)
 #Relationships between tables
Purpose:
 #Metadata helps with organization, searching, and management of information. It makes data easier to find, understand, and use correctly.
```

```python
# Display a bar chart of total_time_watched by platform
display(df[['platform', 'total_time_watched']])
```

```python
# Display a bar chart of total_time_watched by platform
display(df.groupby('platform', as_index=False)['total_time_watched'].sum())
# In the Databricks visualization editor, select "Bar" or "Line" as the chart type.
```

```python
# Display a line chart of total_time_watched by platform using matplotlib
import matplotlib.pyplot as plt

grouped = df.groupby('platform', as_index=False)['total_time_watched'].sum()
plt.figure(figsize=(8, 5))
plt.plot(grouped['platform'], grouped['total_time_watched'], marker='o')
plt.xlabel('Platform')
plt.ylabel('Total Time Watched')
plt.title('Total Time Watched by Platform')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Display a pie chart of total_time_watched by platform
display(df.groupby('platform', as_index=False)['total_time_watched'].sum())
# In the Databricks visualization editor, select "Pie" as the chart type.
# Set "platform" as the key and "total_time_watched" as the value.
```

```python
# Display a pie chart of total_time_watched by platform using matplotlib
import matplotlib.pyplot as plt

grouped = df.groupby('platform', as_index=False)['total_time_watched'].sum()
plt.figure(figsize=(7, 7))
plt.pie(grouped['total_time_watched'], labels=grouped['platform'], autopct='%1.1f%%', startangle=140)
plt.title('Total Time Watched by Platform')
plt.axis('equal')
plt.show()
```
